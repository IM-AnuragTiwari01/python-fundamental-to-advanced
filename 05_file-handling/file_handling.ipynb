```python
# case1: if file is not present

f = open('sample.txt', 'w') # if file does not exists it will getb created in same directrory as we havent provided any location
content: str = """
hi my name is
Anurag Tiwari
i am 
22
years old
"""
f.write(content)
f.close()
```

```python
# case 1.1 : file not present and we want to create filen in another directory
from pathlib import Path
dir = Path.cwd().parent / "test-file"
f = open(dir/"sample1.txt", 'w')
f.write("Hello to new location")
f.close()
```

```python
dir = Path.cwd().parent / "test-file"
f = open(dir/"sample2.txt", 'w')
f.write("Hello to new location")
f.write("\nI am anurag")
f.write("\ni am writting to this file")
f.close()
```

```python
# case 2 : if file already present
dir = Path.cwd().parent / "test-file"
f = open(dir/ "sample.txt", "w")
f.write("Salman Khan")
f.close()
```

```python
# case: 2.1: writting but not overwritting old content
dir = Path.cwd().parent / "test-file"
f = open(dir/ "sample.txt", "a") # opening in append mode
f.write(" is the anchor in bigg boss")
f.write("\nweekend ka vaar")
f.close()
```
