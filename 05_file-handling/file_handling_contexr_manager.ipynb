Using Context manager (with)
- it's a good idea to close a file after usage as it free up the resources
- if we dont close it, garbage collector would close it
- with keyword closes the file as soon as the usage is over

```python
from pathlib import Path
dir = Path.cwd().parent / "test-file"
with open(dir/ "sample1.txt", 'w') as f:
    f.write("pew pew Bhai")    
```

```python
from pathlib import Path
dir = Path.cwd().parent / "test-file"
with open(dir/'sample.txt', 'r') as f:
    print(f.read(11))
    # to know the nextr character or where the buffer is
    print(f.tell())
    f.seek(0)
    print(f.tell())
    print(f.read(11))
```
