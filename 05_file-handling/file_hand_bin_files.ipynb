```python
# Serialising and Deserialising custom objects

class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.gender = gender
    
    def __str__(self) -> str:
        return f" -> {self.fname} -> {self.fname} -> {self.age} -> {self.gender}"
```

```python
person = Person("Anurag", "Tiwari", 33, "Male")
print(person)
```

```python
import json

def show_object(person):
    if isinstance(person, Person):
        return "{} {} age -> {} gender -> {}".format(person.fname, person.lname, person.age, person.gender)

    """
    json.dumps doesn't know how to serialize the custom object so we 
    need to tell it
    """
with open(dir/'person_class_custom_obj.txt', 'w') as f:
    json.dump(person, f, default=show_object)
```
