```python
from pathlib import Path
dir = Path.cwd().parent / "test-file"
f = open(dir/ "sample3.txt", 'r')
s = f.read()
print(s)
f.close()
```

```python
from pathlib import Path
dir = Path.cwd().parent / "test-file"
f = open(dir/ "sample3.txt", 'r')
s = f.read(10) # in argument, can pass how many charaters want to read
print(s)
f.close()
```

```python
# how can we read it line by line
from pathlib import Path
dir = Path.cwd().parent / "test-file"
f = open(dir/ "sample3.txt", 'r')
print(f.readline(), end="")
print(f.readline(), end="")
print(f.readline(), end="")
print(f.readline(), end="")
f.close()
```

```python
#  reading entire file using readline()
from pathlib import Path
dir = Path.cwd().parent / "test-file"
f = open(dir/ "sample3.txt", 'r')
for line in f:
    print(line, end="")
f.close()
```
